# 파이토치 버전 살펴보기

```python
from torchvision.datasets import FashionMNIST

fm_train = FashionMNIST(root='.', train=True, download=True)
fm_test = FashionMNIST(root='.', train=False, download=True)
```

```python
type(fm_train.data)
```

```python
print(fm_train.data.shape, fm_test.data.shape)
```

```python
print(fm_train.targets.shape, fm_test.targets.shape)
```

```python
train_input = fm_train.data
train_target = fm_train.targets
```

```python
# 입력 정규화
train_scaled = train_input / 255.0
```

```python
from sklearn.model_selection import train_test_split

train_scaled, val_scaled, train_target, val_target = train_test_split(
    train_scaled, train_target, test_size=0.2, random_state=42
)
```

```python
print(train_scaled.shape, val_scaled.shape)
```

```python
import torch.nn as nn

model = nn.Sequential(
    nn.Flatten(),
    nn.Linear(784, 100),
    nn.ReLU(),
    nn.Linear(100, 10)
)
```

```python
!pip install torchinfo
```

```python
from torchinfo import summary

summary(model, input_size=(32, 28, 28))
```

```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
```

```python
import torch.optim as optim

criterion = nn.CrossEntropyLoss()  # criterion: 기준
optimizer = optim.Adam(model.parameters())
```

```python
epochs = 5
batches = int(len(train_scaled) / 32)
for epoch in range(epochs):
    model.train()  # 모델을 훈련 모드로 설정한다.
    train_loss = 0  # 현재 에포크의 전체 손실을 저장할 변수를 0으로 초기화한다.
    for i in range(batches):
        inputs = train_scaled[i*32:(i+1)*32].to(device)  # 입력 데이터 32개를 꺼내 GPU/CPU로 이동한다.
        targets = train_target[i*32:(i+1)*32].to(device)  # 해당 입력의 정답 32개를 GPU/CPU로 이동한다.
        optimizer.zero_grad()  # 이전 배치에서 계산된 gradient를 0으로 초기화한다.
        outputs = model(inputs)  # 입력 데이터를 모델에 넣어 예측값을 계산한다.
        loss = criterion(outputs, targets)  # 예측값과 실제 정답을 비교하여 손실을 계산한다.
        loss.backward()  # 손실을 기준으로 각 가중치의 gradient를 계산한다.
        optimizer.step()  # 계산된 gradient를 이용해 가중치를 업데이트한다.
        train_loss += loss.item()  # 현재 배치의 손실값을 누적한다.
    print(f"에포크:{epoch + 1}, 손실:{train_loss/batches:.4f}")
```

```python
model.eval()
with torch.no_grad():
    val_scaled = val_scaled.to(device)
    val_target = val_target.to(device)
    outputs = model(val_scaled)
    predicts = torch.argmax(outputs, 1)
    corrects = (predicts == val_target).sum().item()

accuracy = corrects / len(val_target)
print(f"정확도: {accuracy:.4f}")
```
